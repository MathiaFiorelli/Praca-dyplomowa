```python
from docx import Document
import re
import pandas as pd
import csv
```

```python
document = Document('Pytania Obrona.docx')
text = "\n".join(p.text for p in document.paragraphs)
```

```python
with open("pytania_obrona.txt", "r", encoding="utf-8") as f:
    pytania = f.read()
pytania2=re.sub(r'(\d+\.)\n', r'\1 ', pytania)
pytania2 = pytania2.split("\n")
```

```python
i=24
q = pytania2[i]
q2 = pytania2[i+1]
print(q)

ans = text.split(pytania2[i])[-1].split(pytania2[i+1])[0].strip()
print(ans)
```

```python
answers = []
for i,p in enumerate(pytania2):
    if i!=len(pytania2)-1:
        q = pytania2[i]
        q2 = pytania2[i+1]
        try:
              ans = text.split(pytania2[i])[1].split(pytania2[i+1])[0].strip()
        except:
             print(q)
    else:
        ans = text.split(pytania2[i])[1].strip()
    answers.append(ans)
```

```python
df = pd.DataFrame(columns=["Question", "Answer"])
df["Question"] = pytania2
df["Answer"] = answers
df['Answer'] = df['Answer'].str.replace("\n", "<br>", regex=False)
df
```

```python
df.to_csv("pytania_obrona.csv", index=False, encoding="utf-8",quoting=csv.QUOTE_ALL)
```
